````python
# Creation of a dictionary for drivers
# Creation of a list for the journeys
# Add of a new driver to the dictionary using while

# Dictionary to store drivers information
drivers = {
    1: {"name": "Ala Parson", "phone": "123456789", "total_journeys": 0},
    2: {"name": "John Petty", "phone": "987654321", "total_journeys": 0}
}

# List to store journey information
journeys = [
    {"journey_id": 1, "driver_id": 1, "client_name": "Alice", "distance": 5.0, "amount": 12.5},
    {"journey_id": 2, "driver_id": 2, "client_name": "Bob", "distance": 10.0, "amount": 25.0}
]


    while True:
        # Get driver details from the user
        driver_id = int(input("Enter Driver ID (0 to stop): "))
        
        # Exit the loop if driver_id is 0
        if driver_id == 0:
            break
        
        # Check if driver ID already exists
        if driver_id in drivers:
            print(f"Driver with ID {driver_id} already exists.")
        else:
            name = input("Enter Driver Name: ")
            phone = input("Enter Driver Phone Number: ")
            # Add new driver to the dictionary
            drivers[driver_id] = {"name": name, "phone": phone, "total_journeys": 0}
            print(f"Driver {name} added successfully!")
    
print("Driver entry stopped.")
print("Updated drivers list:")
print(drivers)

````
