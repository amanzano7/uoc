```python
# Paso 1: Crea una variable edad con tu edad como un entero
name = input("Ingresa tu nombre: ")

# Paso 2: Validar la entrada de edad
while True:
    try:
        edad = int(float(input("Ingresa tu edad: ")))  # Convertir a float primero y luego a int
        break  # Salir del bucle si la conversión es exitosa
    except ValueError:
        print("Por favor, ingresa un número entero válido.")

# Paso 3: Convierte la variable edad a una string
edad_str = str(edad)

# Paso 4: Crea el mensaje
mensaje = f"Mi nombre es {name} y mi edad es: {edad_str}"
print(mensaje)
```
